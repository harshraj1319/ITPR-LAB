```python
# 1. Create a list of 15 numbers (taking input from user)
print("Enter 15 numbers separated by space:")
numbers = list(map(int, input().split()))

# Ensure we have at least some numbers to work with
if len(numbers) > 0:
    # 2. Ask user for the number to search/modify
    search_element = int(input("Enter the number to search and remove duplicates of: "))

    # 3. Check if the element exists in the list
    if search_element in numbers:
        # Find the index of the first occurrence
        first_idx = numbers.index(search_element)

        # 4. Create a new list keeping the first occurrence and removing others
        result = []
        found_first = False

        for i in range(len(numbers)):
            if numbers[i] == search_element:
                if not found_first:
                    # Keep the first one we find
                    result.append(numbers[i])
                    found_first = True
                else:
                    # Skip all subsequent occurrences
                    continue
            else:
                # Keep all other different numbers
                result.append(numbers[i])

        print(f"Updated list: {result}")
    else:
        print("Element not found in the list.")
else:
    print("The list is empty.")
```
